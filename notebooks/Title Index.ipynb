```python
from birddog.wiki import ARCHIVES, WIKI_NAMESPACE
from birddog.core import PageLRU, TitleIndex
```

```python
page_lru = PageLRU()
```

```python
title_index = TitleIndex(page_lru=page_lru)
```

```python
for root, addresses in title_index._archives.items():
    for address in addresses:
        page = page_lru.lookup(*address)
        print(root, page.title)
```

```python
for title, address in title_index._index.items():
    page = page_lru.lookup(*address)
    print(title, page.title)
    if not title == f"{WIKI_NAMESPACE}:{page.title}":
        print("MISMATCH")
```

```python
for title in title_index._index.keys():
    print(title, title_index.lookup(title))
```
