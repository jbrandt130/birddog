```python
from birddog.database import Database
```

```python
db = Database()
```

```python
recs = db.scan_all("Locations", view_name="In Modern Ukraine")
```

```python
recs[0]
```

```python
loc_by_id = {}
for rec in recs:
    loc_id = rec["location_id"]
    loc_list = loc_by_id.get(loc_id, [])
    loc_list.append(rec)
    loc_by_id[loc_id] = loc_list    
```
