```python
def _check_finished(self, subtask):
    try:
        self._key_value_store.get(self._completed_id(subtask["task_id"]), str(subtask["index"]))
        return True
    except KeyError:
        pass
    try:
        self._key_value_store.get(self._failed_id(subtask["task_id"]), str(subtask["index"]))
        return True
    except KeyError:
        return False
```
