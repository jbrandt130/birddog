```python
from birddog.database import Database
```

```python
db = Database()
```

```python
p1_docs = []
cursor = None
while True:
    recs, cursor = db.scan(
        "Documents", 
        #view_name="P1", 
        where=("process_code","eq","P1"),
        cursor=cursor, 
        limit=1000, 
        fields=["title", "owning_pages"])
    if not recs:
        break
    p1_docs.extend(recs)
    print(len(p1_docs))
    if not cursor:
        break
```
