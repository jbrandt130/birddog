```python
from birddog.store import SQLiteKeyValueStore, DynamoDBKeyValueStore
from birddog.tracker import (
    _TRACKER_KV_TABLE,
    _TRACKER_KV_NS_TRACKER,
    _TRACKER_KV_NS_CHANGE_LOG,
    )
```

```python
local_store = SQLiteKeyValueStore(table_name=_TRACKER_KV_TABLE)
```

```python
ddb_store = DynamoDBKeyValueStore(table_name=_TRACKER_KV_TABLE)
```

```python
namespaces = [ 
    _TRACKER_KV_NS_TRACKER,
    _TRACKER_KV_NS_CHANGE_LOG,
]
```

```python
items = {}
```

```python
for ns in namespaces:
    print(f"getting items for namespace {ns}...")
    items[ns] = local_store.get_all(ns)
    print(f"...done. got {len(items[ns])} items.")
```
