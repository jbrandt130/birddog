```python
def report_links(links):
    for key, value in links.items():
        value = [item for item in value if "familysearch" not in item]
        if value:
            print(f"  {key}: {value}")
```

```python
def process_title(title):
    address = lookup_page_by_title(title)
    print("=====", title, address)
    if address:
        page = lru.lookup(*address)
        print(page.report)
        print("link:", page.url)
        print("dates:", page._page["dates"])
        print("doc url:", page.doc_url)
        print("other links:")
        report_links(page._page["other_links"])
        print("notes:")
        report_links(page._page["notes"])
        
```

```python
process_title(titles[0])
```

```python
for title in titles:
    process_title(title)
```

```python
def test_lookup(title):
    address = lookup_page_by_title(title)
    if not address:
        print("Failed lookup:", title)
    else:
        page = lru.lookup(*address)
        if not page:
            print("Failed page instantition:", title, address)
```

```python
for title in titles:
    test_lookup(title)
```
