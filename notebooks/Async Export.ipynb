```python
from birddog.runtime import Runtime
from birddog.excel import ExportManager
```

```python
runtime = Runtime()
manager = ExportManager(runtime)
```

```python
title = 'Архів:ДАХмО/Р-6193/12/10423–16825'
template = 'opus.xlsx'
table_name = 'Table 1'
column_map = {
    'ID': [1],
    'DESCRIPTION': [2],
    'DATE': [3]
}
```

```python
id = manager.export_page(
    title, 
    table_name=table_name, 
    template=template, 
    column_map=column_map)
```

```python
manager.is_complete(id)
```

```python
output = manager.get_result(id)
```

```python
id = manager.export_page(
    title,
    compare="2025,01,16,13:29",
    table_name=table_name, 
    template=template, 
    column_map=column_map)
```

```python
manager.is_complete(id)
```

```python
output = manager.get_result(id)
```
