```python
from birddog.task import TaskManager
```

```python
import random
import time
class TaskTest(TaskManager):
    def __init__(self):
        super().__init__("Test")
        self.start()

    def execute_subtask(self, subtask):
        if random.uniform(0, 1) > .2:
            raise ValueError(f"failed on {subtask['payload']}")
        time.sleep(random.uniform(0, 1))
        print("subtask:", subtask)

    def complete_task(self, task_desc, subtasks):
        print("complete:", task_desc, subtasks)
```

```python
mgr = TaskTest()
```

```python
mgr.create("a", list(range(20)))
```

```python
while True:
    print(mgr.active_tasks())
    if not mgr.active_tasks():
        break
    time.sleep(1)
```
